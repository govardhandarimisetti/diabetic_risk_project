```python
df['smoking_status'] = df['smoking_status'].fillna(df.groupby('stress_level')['smoking_status'].transform(lambda x: x.mode()[0] if not x.mode().empty else 'never'))
```

```python
df.isnull().sum()
```

```python
df['alcohol_consumption'] = df['alcohol_consumption'].fillna('unrecorded')
```

```python
df.isnull().sum()
```

```python
df['income_bracket'] = df['income_bracket'].fillna(df.groupby('city')['income_bracket'].transform(lambda x: x.mode()[0] if not x.mode().empty else 'not_specified'))
```

```python
df.isnull().sum()
```

```python
df = df.apply(lambda x: x.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_') if x.dtype == "object" else x)
```

```python
df.columns
```

```python
df = df.drop_duplicates(keep='first')
```

```python
df = df.rename(columns={'hba1c_level': 'avg_blood_sugar'})
```
